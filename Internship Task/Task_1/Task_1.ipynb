```python
#Task 1 
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt 


data = pd.read_csv(r'C:\Users\Hp\Downloads\Iris.csv')
print(data.head())
```

```python
#Adding Shape: rows and columns
print("Shape:", data.shape)

#Writing Names of the Columns:
print("Columns:",data.columns.tolist())

#Data Types
print("Data Types:")
print(data.dtypes)

#Statistically Summary
print("Statistically Summary:")
print(data.describe())
```

```python
#Scatter Plot

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=data,           
    x='PetalLengthCm', 
    y='PetalWidthCm', 
    hue='Species', 
    palette='Set2', 
    s=70, 
    alpha=0.8
)

plt.title('Petal Length vs Petal Width by Species')
plt.xlabel('Petal Length (cm)')  
plt.ylabel('Petal Width (cm)')   
plt.legend(title='Species')
plt.tight_layout()
plt.show()
```

```python
#Histogram

sns.displot(
    data=data,
    x='PetalLengthCm',
    hue='Species',
    kind='hist',
    kde=True,
    bins=20,
    palette='Set2',
    height=4,
    aspect=1.8,
    alpha=0.5
)

plt.title('Distribution of Petal Length by Species')
plt.xlabel('Petal Length (cm)')
plt.tight_layout()
plt.show()
```

```python
#Box Plot
# Melt to long format first
data_melted = data.melt(
    id_vars='Species',
    value_vars=['SepalLengthCm', 'SepalWidthCm', 'PetalLengthCm', 'PetalWidthCm'],
    var_name='Feature',
    value_name='Value'
)

plt.figure(figsize=(11, 5))

sns.boxplot(
    data=data_melted,
    x='Feature',
    y='Value',
    hue='Species',
    palette='Set2',
    width=0.6
)

plt.title('All Features — Spread and Outliers by Species')
plt.xlabel('Feature')
plt.ylabel('Value (cm)')
plt.legend(title='Species', bbox_to_anchor=(1.01, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
